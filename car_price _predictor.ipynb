# Cleaning

```python
car['year'] = car['year'].astype(int)
```

```python
car = car[car['Price']!="Ask For Price"]
```

```python
car['Price'] = car['Price'].str.replace(',','').astype(int)
```

```python
car
```

```python
car['kms_driven'] = car['kms_driven'].str.split(' ').str.get(0).str.replace(',','')
```

```python
car = car[car['kms_driven'].str.isnumeric()]
```

```python
car
```

```python
 car['kms_driven'] = car['kms_driven'].astype(int)
```

```python
car
```

```python
car = car[~car['fuel_type'].isna()]
```

```python
car
```

```python
car['name']=car['name'].str.split(' ').str.slice(0,3).str.join(' ')
```

```python
car = car.reset_index(drop=True)
```

```python
car = car.drop('Unnamed: 0',axis = 1)
```

```python
car = car[car['Price']<6e6].reset_index(drop = True)
```

# Model

```python
x = car.drop(columns = 'Price')
y = car['Price']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test = train_test_split(x,y,test_size = 0.2)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import make_column_transformer
from sklearn.pipeline import make_pipeline
```

```python
one =  OneHotEncoder()
one.fit(x[['name','company','fuel_type']])
```

```python
colum_Trans = make_column_transformer((OneHotEncoder(categories  = one.categories_),['name','company','fuel_type']),remainder = 'passthrough')
```

```python
lr = LinearRegression()
```

```python
pipe = make_pipeline(colum_Trans,lr)
```

```python
pipe.fit(X_train,Y_train)
```

```python
Y_pred = pipe.predict(X_test)
```

```python
r2_score(Y_test,Y_pred)
```

```python
scores = []
for i in range(1000):
    X_train,X_test,Y_train,Y_test = train_test_split(x,y,test_size = 0.2,random_state = i)
    lr = LinearRegression()
    pipe = make_pipeline(colum_Trans,lr)
    pipe.fit(X_train,Y_train)
    Y_pred = pipe.predict(X_test)
    scores.append(r2_score(Y_test,Y_pred))
```

```python
np.argmax(scores)
```

```python
scores[np.argmax(scores)]
```

```python
for i in range(10):
     X_train,X_test,Y_train,Y_test = train_test_split(x,y,test_size = 0.2,random_state = np.argmax(scores))
     lr = LinearRegression()
     pipe = make_pipeline(colum_Trans,lr)
     pipe.fit(X_train,Y_train)
     Y_pred = pipe.predict(X_test)
     print(r2_score(Y_test,Y_pred),i)
```
